## 2. Planteamiento del Problema

Desde la masificación de la telefonía celular y los equipos portátiles, los hábitos horarios de la población general han experimentado un desplazamiento nocturno sostenido (LeBourgeois et al., 2017). Este patrón se acentúa en la población estudiantil universitaria de 18 a 29 años, donde el uso diario de teléfonos inteligentes supera el 96 % (Pew Research Center, 2021) y la interacción con pantallas se prolonga frecuentemente dentro de la cama en la franja nocturna (Maurya et al., 2022).

La interacción recurrente con estos equipos en horarios nocturnos altera la arquitectura del descanso mediante mecanismos fisiológicos y conductuales documentados en la literatura (LeBourgeois et al., 2017): la sustitución del tiempo destinado a dormir por el consumo de contenidos digitales, la estimulación neurocognitiva derivada de las pantallas interactivas y la exposición directa a la luz azul, la cual inhibe la secreción de melatonina y retrasa los ritmos circadianos. Esta alteración compromete el descanso regular requerido para el rendimiento intelectual y la salud integral (Hirshkowitz et al., 2015). A escala mundial, cerca de un tercio de la población adulta duerme por debajo del rango óptimo de 7 a 9 horas diarias (Liu et al., 2016).

En la Escuela de Estadística y Ciencias Actuariales de la Universidad Central de Venezuela (EECA-UCV), los planes de estudio exigen una alta carga cognitiva, elevado esfuerzo de abstracción cuantitativa y jornadas prolongadas frente al computador en tareas de programación, cálculo actuarial y análisis de datos. Durante el período académico 2026-01S, la comunidad estudiantil asume compromisos académicos y traslados urbanos que desplazan frecuentemente las horas de lectura, estudio e interacción tecnológica hacia **horarios poco convencionales**. Bajo un criterio cronobiológico (Borbély, 1982; Hirshkowitz et al., 2015), esta franja abarca entre las 22:00 y las 06:00 horas, incluyendo las dos a tres horas previas a dormir donde el brillo de la pantalla frena la producción pineal de melatonina.

Aunque el desgaste matutino suele reportarse entre el alumnado durante las primeras horas de clase, la EECA-UCV carece de un levantamiento estadístico formal que describa la persistencia de estos hábitos y cuantifique la asociación empírica entre la exposición tecnológica previa al reposo y el déficit de sueño en su población estudiantil. Ante este vacío de información empírica, se formula la siguiente pregunta central de investigación:

¿De qué manera la duración de la exposición nocturna a pantallas de dispositivos electrónicos altera la calidad y las horas efectivas de sueño en los estudiantes de la EECA-UCV?

## 3. Objetivos

### 3.1. Objetivo General

Analizar la relación lineal entre el tiempo de uso nocturno de dispositivos electrónicos y la calidad del sueño en los estudiantes de la EECA-UCV durante el período 2026-01S, mediante un estudio observacional por muestreo.

### 3.2. Objetivos Específicos

1. **Caracterizar** los hábitos de consumo digital nocturno según el tipo de dispositivo, la actividad predominante y la ventana de exposición previa al descanso.
2. **Estimar** la duración absoluta y la calidad percibida del sueño durante los días lectivos frente a los fines de semana.
3. **Identificar** las variaciones en los patrones de descanso asociadas a los distintos niveles de exposición a pantallas.

## 5. Justificación

El descanso nocturno cumple un proceso neurobiológico activo indispensable para la consolidación sináptica, la memoria de trabajo y la recuperación metabólica (Hirshkowitz et al., 2015). En la EECA-UCV, donde el rendimiento estudiantil exige sostener la concentración ante asignaturas de alta densidad analítica como Matemáticas, Estadística y Probabilidades, la restricción del sueño deteriora la agilidad cognitiva y eleva el margen de error cuantitativo durante las semanas de evaluación.

En el plano metodológico, la investigación entrega a la comunidad académica un instrumento de medición validado, una matriz de operacionalización estandarizada y un protocolo de depuración de sesgos horarios adaptado al estudiante universitario venezolano. Futuros estudios observacionales en otras facultades podrán replicar esta estructura de medición para contrastar la asimetría semanal del descanso y la disrupción circadiana entre distintas disciplinas.

## 8. Antecedentes

La base empírica de esta investigación parte de cinco estudios internacionales recientes que examinaron la relación entre el uso de tecnología digital y distintos parámetros del descanso nocturno.

En 2025, He, Pan, Ma, Li, Kong, Liu, Liu, Wang, Deng y Yang publicaron la revisión sistemática y metaanálisis titulada *The association of screen time and the risk of sleep outcomes: A systematic review and meta-analysis* en *Frontiers in Psychiatry*, volumen 16, artículo 1640263, reuniendo 21 estudios de cohorte con 548.338 participantes. A partir de cuestionarios longitudinales y registros de sueño procesados con modelos de metarregresión y razones de probabilidad agrupadas, calcularon que cada hora diaria adicional frente a pantallas resta entre 3 y 5 minutos de sueño con $\beta = -0{,}05$, eleva el riesgo de sueño corto con $OR = 1{,}25$, triplica la dificultad para conciliarlo con $OR = 3{,}05$ y atrasa la hora de acostarse en $13{,}2$ minutos por hora. Esos coeficientes fijan la pérdida marginal esperada que guía nuestras hipótesis documentales, además de respaldar en el plano metodológico la regresión lineal de los minutos continuos de pantalla sobre las horas dormidas.

En *Sleep and use of electronic devices in adolescence: Results from a large population-based study*, publicado en *BMJ Open*, volumen 5, número 1, e006748, Hysing, Pallesen, Stormark, Jakobsen, Lundervold y Sivertsen encuestaron en 2015 por autorreporte a 9.846 jóvenes noruegos de 16 a 19 años sobre el uso de pantallas en la cama y sus horarios de descanso. Sus regresiones logísticas ajustadas por variables sociodemográficas mostraron que usar computadoras o teléfonos antes de dormir eleva la probabilidad de dormir menos de 5 horas, con $OR = 2{,}70$ al usar computador al acostarse y $OR = 3{,}64$ al superar 4 horas diarias totales, además de aumentar en 52 % y 48 % la razón de momios de presentar un déficit superior a 2 horas.

De ese antecedente se adopta la solución metodológica de preguntar por separado según el tipo de dispositivo principal y estimar las horas dormidas mediante la diferencia horaria entre acostarse y levantarse.

En el artículo *Digital media and sleep in childhood and adolescence* de la revista *Pediatrics*, volumen 140, suplemento 2, páginas S92 a S96, LeBourgeois, Hale, Chang, Akacem, Montgomery-Downs y Buxton examinaron en 2017 trabajos transversales, longitudinales y experimentales apoyados en cuestionarios digitales, actigrafía y mediciones fisiológicas de fase circadiana. Sus resultados mostraron que el consumo nocturno de pantallas retrasa la hora de acostarse y acorta el descanso por desplazamiento del tiempo disponible, activación psicológica y supresión de melatonina debido a la luz de onda corta. En la presente investigación cumple una función estrictamente documental, ya que articula el modelo regulador de dos procesos empleado para interpretar las interrupciones nocturnas y la fatiga al despertar.

En 2022, Maurya, Muhammad, Maurya y Dhillon divulgaron *The association of smartphone screen time with sleep problems among adolescents and young adults: Cross-sectional findings from India* en *BMC Public Health*, volumen 22, número 1, artículo 1686, procesando los datos transversales de 16.292 adolescentes y adultos jóvenes de la cohorte UDAYA. Con tablas de contingencia, pruebas chi-cuadrado y regresión logística multivariable, detectaron trastornos del sueño en el 18,4 % de los hombres y el 33,24 % de las mujeres, verificando que exceder las dos horas diarias en el teléfono incrementa la probabilidad de sufrir problemas del sueño con $AOR = 1{,}48$ e intervalo de confianza del 95 % entre $1{,}24$ y $1{,}75$.

La decisión metodológica de agrupar la exposición en tramos horarios críticos de 120 y 180 minutos o más y contrastar las categorías mediante pruebas $\chi^2$ en el cruce bivariado proviene directamente de este diseño transversal.

En 2025, Zhong, Masters, Donzella, Diver y Patel analizaron en *Electronic screen use and sleep duration and timing in adults*, publicado en *JAMA Network Open*, volumen 8, número 3, e252493, las respuestas de 122.058 adultos de la cohorte CPS-3 mediante razones de prevalencia ajustadas y ecuaciones de estimación generalizada sobre cuestionarios de cronotipo y pantallas en la hora previa al descanso. Encontraron que el 41,2 % utilizaba pantallas a diario antes de dormir, con 33 % mayor prevalencia de mala calidad del sueño, $RP = 1{,}33$ e intervalo al 95 % de $1{,}27$ a $1{,}39$, además de una merma cercana a 50 minutos semanales, más pronunciada en días laborables con $-7{,}64\text{ min/noche}$ que en días libres con $-5{,}04\text{ min/noche}$. Documentalmente respalda el contraste entre días de trabajo y descanso, mientras que en el plano metodológico justifica haber separado en nuestro formulario los horarios de lunes a viernes de los de sábado y domingo.

## 9. Marco Teórico Referencial

### 9.1. Modelo de Dos Procesos de Regulación del Sueño (Borbély, 1982)

La fisiología del descanso humano se rige por la interacción de dos fuerzas reguladoras:

1. **Proceso C de regulación circadiana:** Oscilación endógena biológica cuasi-sinusoidal con un período aproximado de 24 horas, generada por el marcapasos central ubicado en el núcleo supraquiasmático (NSQ) del hipotálamo anterior. Este mecanismo sincroniza las ventanas de alerta y propensión al sueño con las señales lumínicas ambientales captadas por las células ganglionares de la retina.
2. **Proceso S o presión homeostática:** Acumulación neuroquímica progresiva de propensión al descanso, mediada por la adenosina en el prosencéfalo basal, generada por cada hora de vigilia continua. Durante el sueño no REM, el Proceso S decae exponencialmente según la ecuación de descarga:
   $$S(t) = S_0 \cdot e^{-t/\tau}$$
   donde $\tau$ representa la constante de depuración homeostática cerebral.

### 9.2. Mecanismos Fisiológicos de Disrupción Tecnológica

El uso nocturno de dispositivos con retroiluminación LED altera ambos procesos mediante tres mecanismos descritos por LeBourgeois et al. (2017).

**Desplazamiento temporal del reposo:** la interactividad digital posterga deliberadamente el momento de apagar la luz e intentar conciliar el sueño, acortando la ventana física disponible para descansar antes del horario de levantarse.

**Activación cognitiva y neuroemocional:** la lectura, las redes sociales y la mensajería mantienen activa la corteza prefrontal, sostienen la frecuencia cardíaca y retrasan la desactivación psicofisiológica necesaria para iniciar el sueño.

**Supresión melatonérgica por luz azul:** las pantallas emiten radiación de onda corta entre $450$ y $480\text{ nm}$ que estimula los fotopigmentos de melanopsina en la retina, inhibiendo la síntesis pineal de melatonina y desplazando la fase del Proceso C.

## 11. Variables en Estudio

### 11.1. Estrategia de Medición Cronológica Indirecta

Para eliminar el sesgo de estimación retrospectiva, donde el informante aproxima erróneamente sus horas dormidas, la investigación adopta una **medición cronológica indirecta**. La duración aproximada del sueño se calcula a partir de la diferencia modular entre la hora de acostarse y la de levantarse:

$$t_{\text{sueño}} = \begin{cases} 
t_{\text{levantarse}} - t_{\text{acostarse}}, & \text{si } t_{\text{levantarse}} \ge t_{\text{acostarse}} \\[0.2cm]
(t_{\text{levantarse}} + 24) - t_{\text{acostarse}}, & \text{si } t_{\text{levantarse}} < t_{\text{acostarse}}
\end{cases}$$

Este cálculo se desagrega para días laborables lectivos $t_{\text{sueño, sem}}$ y fines de semana $t_{\text{sueño, fin}}$, permitiendo cuantificar la compensación o *jetlag social*:
$$\Delta t_{\text{jetlag}} = t_{\text{sueño, fin}} - t_{\text{sueño, sem}}$$

```python
# ==============================================================================
# PIPELINE DE PROCESAMIENTO, FRECUENCIAS Y DESCRIPTIVOS (n = 28)
# ==============================================================================
import pandas as pd
import numpy as np
from scipy import stats

ARCHIVO = "Encuesta sobre uso de dispositivos y calidad del sueño (Respuestas).xlsx"
df_raw = pd.read_excel(ARCHIVO).iloc[:28].copy()
df2 = df_raw.drop(columns=["Marca temporal"]).copy()
df2.columns = [f"Pregunta {i}" for i in range(1, 13)]

for c in df2.select_dtypes(include=["object", "string"]).columns:
    df2[c] = df2[c].map(lambda v: v.strip() if isinstance(v, str) else v)

df2["Edad"] = pd.to_numeric(
    df2["Pregunta 2"].astype(str).str.extract(r"(\d+)")[0], errors="coerce"
).astype("int64")

def to_hours(val):
    if hasattr(val, "hour"):
        return val.hour + val.minute / 60.0
    h, m = str(val).split(":")[:2]
    return int(h) + int(m) / 60.0

def fix_bedtime(h):
    h_arr = np.asarray(h, dtype=np.float64)
    return np.where(
        (h_arr >= 8.0) & (h_arr < 12.0),
        h_arr + 12.0,
        np.where((h_arr >= 12.0) & (h_arr < 13.0), h_arr - 12.0, h_arr),
    )

lev_sem = df2["Pregunta 7"].apply(to_hours).to_numpy(dtype=np.float64)
aco_sem = fix_bedtime(df2["Pregunta 8"].apply(to_hours))
lev_fin = df2["Pregunta 9"].apply(to_hours).to_numpy(dtype=np.float64)
aco_fin = fix_bedtime(df2["Pregunta 10"].apply(to_hours))

sem = np.where(aco_sem >= 14.0, (24.0 - aco_sem) + lev_sem, lev_sem - aco_sem)
fin = np.where(aco_fin >= 14.0, (24.0 - aco_fin) + lev_fin, lev_fin - aco_fin)
inconsistente = (df2["Pregunta 8"].apply(to_hours).to_numpy(dtype=np.float64) == lev_sem) | (sem <= 0)
sem[inconsistente] = np.sort(sem[~inconsistente])[0]  # Cota mínima válida = 4.00 h

df2["Sueno_Semana"] = sem
df2["Sueno_Finde"] = fin
df2["Sueno_Ponderado"] = (5.0 * sem + 2.0 * fin) / 7.0
df2["Compensacion_Finde"] = fin - sem

mapa_minutos = {
    "Menos de 30 minutos": 15.0,
    "De 30 a 59 minutos": 44.5,
    "De 60 a 119 minutos": 89.5,
    "120 a 179 minutos": 149.5,
    "180 minutos o más": 195.0,
}
mapa_calidad = {
    "Somnolencia persistente que afecta su concentración en clases.": 1,
    "Cansancio leve que desaparece tras la primera hora.": 2,
    "Energía suficiente para toda la jornada académica.": 3,
}
df2["Minutos_Pantalla"] = df2["Pregunta 5"].map(mapa_minutos)
df2["Calidad_Sueño"] = df2["Pregunta 12"].map(mapa_calidad)

print(f"Observaciones válidas procesadas: n = {len(df2)}")
print("Descriptivos horas de sueño (semana / finde / ponderado / delta):")
print(df2[["Sueno_Semana", "Sueno_Finde", "Sueno_Ponderado", "Compensacion_Finde"]].describe().round(2))
```

### 14.2. Interpretación y Análisis de Resultados

Los datos de los 28 estudiantes muestran tres resultados principales:

1. **Prevalencia de dispositivos y tiempo de exposición:**  
En el Cuadro 1 se observa que el 67,86 % de los estudiantes de la Escuela utilizan teléfonos inteligentes horas previas al descanso. #### 14.3.1. Estimación por Intervalos

Con la muestra de 28 estudiantes de la EECA se calcularon intervalos de confianza del 95 % para los parámetros clave empleando la distribución $t$ de Student con 27 grados de libertad debido al tamaño reducido de la muestra.  
La media del sueño en días lectivos arrojó un intervalo de $\mu_{\text{sem}}$ entre $6{,}08$ y $7{,}26$ horas; el límite inferior muestra que buena parte del grupo duerme cerca de las 6 horas.  
La media del sueño en fines de semana arrojó un intervalo de $\mu_{\text{fin}}$ entre $8{,}32$ horas y $9{,}54$ horas, el cual no presenta intersección con el intervalo anterior en los días de semana lectiva; como ambos promedios provienen de los mismos 28 estudiantes, el cambio real se evalúa con el intervalo de la diferencia individual $\Delta t$.  
La media de compensación horaria se encontró en un intervalo de $1{,}42$ horas a $3{,}10$ horas. Al excluir el cero, indica que efectivamente la muestra estudiantil posee una deuda de sueño durante la semana, la cual es compensada los días de fines de semana.


```python
# ==============================================================================
# ESTIMACIÓN POR INTERVALOS DE CONFIANZA (95 %) Y NORMALIDAD (SHAPIRO-WILK)
# ==============================================================================
def intervalo_media(data, confianza=0.95):
    a = np.asarray(data, dtype=float)
    n = len(a)
    m, se = np.mean(a), stats.sem(a)
    h = se * stats.t.ppf((1.0 + confianza) / 2.0, n - 1)
    return m, m - h, m + h

for etiqueta, col in [
    ("Sueño Semana (h)", "Sueno_Semana"),
    ("Sueño Fin de Semana (h)", "Sueno_Finde"),
    ("Compensación Finde (h)", "Compensacion_Finde"),
]:
    m, li, ls = intervalo_media(df2[col])
    w, p_sw = stats.shapiro(df2[col])
    print(f"{etiqueta}: media = {m:.4f}, IC 95% = [{li:.2f}, {ls:.2f}] | Shapiro-Wilk W = {w:.4f}, p = {p_sw:.4f}")
```

#### 14.3.2. Contraste de Hipótesis

Se realizaron 4 pruebas de contraste de hipótesis en esta investigación, resumidas en el Cuadro 10 al 5 % de significancia, las cuales son expuestas a continuación:  

1. **Prueba de Hipótesis 1 correspondiente al déficit de sueño respecto al umbral saludable:**
   $$H_0: \mu_{\text{sem}} \ge 7 \quad \text{vs.} \quad H_1: \mu_{\text{sem}} < 7$$
   Se obtuvo un $t = -1{,}1559$ y $p = 0{,}1289$, por lo que no se rechaza la hipótesis nula. A pesar de que la medida es deficitaria, la variabilidad en la muestra y el tamaño de la misma limitan la potencia del contraste de hipótesis para certificar el déficit generalizado al 95 %.

2. **Prueba de Hipótesis 2 sobre asimetría semanal y fenómeno del Jetlag Social:**
   $$H_0: \mu_{\Delta t} \le 0 \quad \text{vs.} \quad H_1: \mu_{\Delta t} > 0$$
   La prueba arroja un $t = 5{,}5401$ con $p = 3{,}58 \times 10^{-6}$, por lo que se rechaza la hipótesis nula y existe una compensación horaria real en términos del descanso en los fines de semana.

3. **Prueba de Hipótesis 3 sobre regresión lineal e independencia en tablas de contingencia:**
   $$H_0: \text{El tiempo de pantalla y la duración o calidad del sueño son independientes}$$
   $$H_1: \text{Existe relación lineal o dependencia estadística entre las variables}$$
   Con un $F = 1{,}494$ y $p = 0{,}2326$ en la regresión lineal del Cuadro 7, junto a $\chi^2 = 5{,}7100$ y $p = 0{,}2219$ en la contingencia del Cuadro 5, no hubo elementos suficientes para rechazar la hipótesis nula, por lo que no se puede confirmar que exista una relación directa entre los minutos de exposición a pantallas de dispositivos electrónicos y el descanso.

4. **Prueba de Hipótesis 4 sobre correlación lineal y monótona:**
   $$H_0: \rho = 0 \quad \text{vs.} \quad H_1: \rho \neq 0$$
   Con $r_s = +0{,}4064$ y $p = 0{,}0318$ en el coeficiente de Spearman del Cuadro 8 se rechaza la hipótesis nula, lo que confirma una asociación monótona estadísticamente significativa de signo positivo entre el tiempo nocturno frente a pantallas y la calidad percibida al despertar en esta muestra.


```python
# ==============================================================================
# MÓDULO DE PRUEBAS DE HIPÓTESIS Y CORRELACIONES (n = 28)
# ==============================================================================
import statsmodels.api as sm

t1 = stats.ttest_1samp(df2["Sueno_Semana"], popmean=7.0, alternative="less")
t2 = stats.ttest_rel(df2["Sueno_Finde"], df2["Sueno_Semana"], alternative="greater")
w2_bi = stats.wilcoxon(df2["Sueno_Finde"], df2["Sueno_Semana"])
w2_uni = stats.wilcoxon(df2["Sueno_Finde"], df2["Sueno_Semana"], alternative="greater")

X = sm.add_constant(df2["Minutos_Pantalla"])
modelo = sm.OLS(df2["Sueno_Semana"], X).fit()

ct_3x3 = pd.crosstab(
    df2["Pregunta 5"].map({
        "Menos de 30 minutos": "< 60 min",
        "De 30 a 59 minutos": "< 60 min",
        "De 60 a 119 minutos": "60 a 119 min",
        "120 a 179 minutos": "120 min o más",
        "180 minutos o más": "120 min o más",
    }),
    df2["Calidad_Sueño"],
)
chi2_val, p_chi2, dof_chi2, _ = stats.chi2_contingency(ct_3x3)

r_pear, p_pear = stats.pearsonr(df2["Minutos_Pantalla"], df2["Sueno_Semana"])
r_sp_cal, p_sp_cal = stats.spearmanr(df2["Minutos_Pantalla"], df2["Calidad_Sueño"])
desp_num = df2["Pregunta 11"].map({"Ninguna": 0, "Una vez": 1, "Dos veces": 2, "3 o más veces": 3})
r_sp_desp, p_sp_desp = stats.spearmanr(df2["Minutos_Pantalla"], desp_num)

print(f"Prueba 1 (mu_sem < 7): t = {t1.statistic:.4f}, gl = 27, p = {t1.pvalue:.4f}")
print(f"Prueba 2 (delta > 0): t = {t2.statistic:.4f}, gl = 27, p_unilateral = {t2.pvalue:.4e} | Wilcoxon W = {w2_bi.statistic:.1f}, p_uni = {w2_uni.pvalue:.4e}")
print(f"Prueba 3a (OLS): F(1, 26) = {modelo.fvalue:.4f}, p = {modelo.f_pvalue:.4f}, R2 = {modelo.rsquared:.4f}, R2_aj = {modelo.rsquared_adj:.4f}")
print(f"Prueba 3b (Chi2 3x3): chi2 = {chi2_val:.4f}, gl = {dof_chi2}, p = {p_chi2:.4f}")
print(f"Prueba 4 (Pearson / Spearman, gl = 26): r = {r_pear:.4f} (p={p_pear:.4f}), rs_cal = {r_sp_cal:+.4f} (p={p_sp_cal:.4f}), rs_desp = {r_sp_desp:.4f} (p={p_sp_desp:.4f})")
```

### Anexo C: Glosario de Términos

**Correlación de Pearson ($r$):** Coeficiente entre -1 y 1 que mide únicamente qué tan recta es la relación entre dos variables continuas.

**Correlación de Spearman ($r_s$):** Medida no paramétrica calculada sobre los rangos ordenados de los datos con $gl = n - 2$. Sirve para cruzar variables continuas con escalas ordinales, como la sensación al despertar, sin exigir distribución normal.

**Higiene del sueño:** Conjunto de hábitos diarios y condiciones del cuarto que permiten descansar de corrido sin alterar el reloj biológico. Incluye soltar el teléfono antes de acostarse, evitar actividad mental intensa en la cama y no desordenar los horarios los fines de semana.

**Imputación por cota mínima:** Reemplazo de un dato horario inválido por el menor valor real observado en la muestra, igual a $4{,}00\text{ h}$, para no perder el registro ni meter ceros ficticios.

**Jetlag social ($\Delta t$):** Horas extra que se duermen el fin de semana frente a los días de clases, $\Delta t = t_{\text{fin}} - t_{\text{sem}}$, para recuperar sueño acumulado.

**Latencia del sueño:** Tiempo que tarda una persona en quedarse dormida desde que apaga la luz.

**Luz azul:** Franja visible de onda corta entre $450$ y $480\text{ nm}$ emitida por pantallas LED que frena la producción nocturna de melatonina al llegar directo a la retina.

**Mínimos Cuadrados Ordinarios (OLS / MCO):** Método de ajuste que estima la recta minimizando la suma de las diferencias verticales al cuadrado, $\sum e_i^2$. Con $n = 28$, la pendiente estimada necesita ser muy marcada para diferenciarse estadísticamente de cero bajo las pruebas habituales.

**Potencia estadística ($1 - \beta$):** Probabilidad de detectar un efecto verdadero sin cometer error tipo II dado el tamaño muestral $n$.

**Proceso C y Proceso S:** Dos mecanismos del modelo propuesto por Borbély en 1982: el ciclo circadiano cercano a 24 horas y la presión homeostática que sube conforme aumentan las horas despierto.

**Prueba de Shapiro-Wilk ($W$):** Contraste de normalidad cuya hipótesis nula supone que los datos siguen una campana normal. Si $p < 0{,}05$, como pasó con $\Delta t$ con $p = 0{,}0456$, conviene acudir a pruebas no paramétricas.

**Prueba de rangos con signo de Wilcoxon:** Contraste pareado sin supuesto de normalidad para comprobar si la mediana de $\Delta t$ se aparta de cero.

**Sesgo de confusión:** Distorsión al interpretar dos variables causada por un tercer factor ligado a ambas.
