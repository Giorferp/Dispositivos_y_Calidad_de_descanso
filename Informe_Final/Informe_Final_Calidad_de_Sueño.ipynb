## 2. Planteamiento del Problema

Desde la adopción masiva de la telefonía móvil y los terminales digitales portátiles, los hábitos horarios de la población general han experimentado un desplazamiento nocturno sostenido (LeBourgeois et al., 2017). Este patrón se acentúa en la población estudiantil universitaria, cuyos integrantes emplean de forma intensiva teléfonos inteligentes, computadoras portátiles y tabletas en sus actividades diurnas y nocturnas.

La interacción recurrente con estos terminales en horarios nocturnos altera la arquitectura del descanso mediante mecanismos fisiológicos y conductuales documentados en la literatura (LeBourgeois et al., 2017): la sustitución del tiempo destinado a dormir por el consumo de contenidos digitales, la estimulación neurocognitiva derivada de las pantallas interactivas y la exposición directa a la luz azul, la cual inhibe la secreción de melatonina y retrasa los ritmos circadianos. Esta alteración compromete el descanso regular requerido para el rendimiento intelectual y la salud integral (Hirshkowitz et al., 2015). A escala mundial, cerca de un tercio de la población adulta duerme por debajo del rango óptimo de 7 a 9 horas diarias (Liu et al., 2016).

En la Escuela de Estadística y Ciencias Actuariales de la Universidad Central de Venezuela (EECA-UCV), los planes de estudio exigen una alta carga cognitiva, elevado esfuerzo de abstracción cuantitativa y jornadas prolongadas frente al computador en tareas de programación, cálculo actuarial y análisis de datos. Durante el período académico 2026-01S, la comunidad estudiantil asume compromisos académicos y traslados urbanos que desplazan frecuentemente las horas de lectura, estudio e interacción tecnológica hacia la noche tardía.

Aunque el cansancio matutino se manifiesta de manera cotidiana en las aulas de clase, la EECA-UCV carece de un relevamiento métrico formal que describa la persistencia de estos hábitos y cuantifique la asociación empírica entre la exposición tecnológica previa al reposo y el déficit de sueño en su alumnado. Ante este vacío de información empírica, se formula la siguiente pregunta central de investigación:

¿De qué manera la duración de la exposición nocturna a pantallas de dispositivos electrónicos altera la calidad y las horas efectivas de sueño en los estudiantes de la EECA-UCV?

## 3. Objetivos

### 3.1. Objetivo General

Analizar la relación lineal entre el tiempo de uso nocturno de dispositivos electrónicos y la calidad del sueño en los estudiantes de la EECA-UCV durante el período 2026-01S.

### 3.2. Objetivos Específicos

1. **Caracterizar** los hábitos de consumo digital nocturno según el tipo de dispositivo y la ventana de exposición previa al descanso.
2. **Estimar** la duración absoluta y la calidad percibida del sueño durante los días lectivos frente a los fines de semana.
3. **Identificar** las variaciones en los patrones de descanso asociadas a los distintos niveles de exposición a pantallas.


---

## 4. Finalidad

La recolección de estos datos primarios permite dimensionar una problemática latente en la rutina universitaria. Cuantificar el déficit de sueño frente a la exposición digital proporciona la base métrica empírica necesaria para que las autoridades de la Escuela estructuren campañas de higiene del sueño fundamentadas en su propia población, sustituyendo las suposiciones genéricas sobre el cansancio estudiantil.


---

## 5. Justificación

El descanso nocturno cumple un proceso neurobiológico activo indispensable para la consolidación sináptica, la memoria de trabajo y la recuperación metabólica (Hirshkowitz et al., 2015). En la EECA-UCV, donde el rendimiento del estudiante exige mantener una concentración sostenida durante largos períodos frente a asignaturas de alta exigencia analítica (Matemáticas, Estadísticas, Probabilidades, entre otras), la privación de sueño deteriora directamente la agilidad mental y eleva el margen de error en el razonamiento cuantitativo, sobre todo durante semanas de alta carga académica.

En el plano metodológico, la investigación aporta a la comunidad académica un instrumento de medición validado para la realidad del estudiante universitario venezolano. El diseño estructural del cuestionario y la operacionalización de las variables establecen un precedente referencial. Futuros estudios observacionales en otras facultades podrán replicar esta arquitectura para contrastar el fenómeno del "jetlag social" y la disrupción circadiana en diferentes disciplinas.

Los autores cuantificaron el riesgo relativo y la reducción marginal de minutos de sueño mediante modelos de metarregresión, lo cual fundamenta nuestra operacionalización del tiempo de exposición en intervalos continuos y la evaluación del riesgo de insomnio matutino.

* **Hysing, M., Pallesen, S., Stormark, K. M., Jakobsen, R., Lundervold, A. J., & Sivertsen, B. (2015):**  
  *Sleep and use of electronic devices in adolescence: Results from a large population-based study*. Publicado en *BMJ Open*, *5*(1), e006748.  
  Este estudio establece un precedente directo sobre el autorreporte de dispositivos en jóvenes. Los autores emplearon modelos de regresión logística ajustados por variables sociodemográficas para estimar el déficit de sueño, lo cual justifica nuestra estrategia de medir la latencia pre-sueño y discriminar el uso según el tipo de pantalla interactiva.

* **LeBourgeois, M. K., Hale, L., Chang, A. M., Akacem, L. D., Montgomery-Downs, H. E., & Buxton, O. M. (2017):**

  Los autores estructuran el desplazamiento temporal, la activación psicológica y la luz azul dentro del modelo de dos procesos, sustentando nuestra medición diferenciada de los horarios de acostarse frente a las horas efectivas de sueño.

* **Maurya, C., Muhammad, T., Maurya, P., & Dhillon, P. (2022):**  
  *The association of smartphone screen time with sleep problems among adolescents and young adults: Cross-sectional findings from India*. Publicado en *BMC Public Health*, *22*(1), 1686.  
  Mediante un diseño transversal con jóvenes universitarios, los autores aplicaron pruebas de chi-cuadrado y razones de prevalencia estratificadas para evaluar el umbral crítico de tres horas, lo cual orienta la construcción de nuestras tablas de contingencia y el análisis bivariado de datos.

* **Zhong, C., Masters, M., Donzella, S. M., Diver, W. R., & Patel, A. V. (2025):**  
  *Electronic screen use and sleep duration and timing in adults*. Publicado en *JAMA Network Open*, *8*(3), e252493.

## 9. Marco Teórico Referencial

### 9.1. Modelo de Dos Procesos de Regulación del Sueño (Borbély, 1982)

La fisiología del descanso humano se rige por la interacción de dos fuerzas reguladoras:

1. **Proceso C (Regulación Circadiana):**  
   Oscilación endógena biológica cuasi-sinusoidal con un período aproximado de 24 horas, generada por el marcapasos central ubicado en el núcleo supraquiasmático (NSQ) del hipotálamo anterior. Este mecanismo sincroniza las ventanas de alerta y propensión al sueño con las señales lumínicas ambientales captadas por las células ganglionares de la retina.

2. **Proceso S (Presión Homeostática):**  
   Acumulación neuroquímica progresiva de propensión al descanso (mediada por la adenosina en el prosencéfalo basal) generada por cada hora de vigilia continua. Durante el sueño no REM, el Proceso S decae exponencialmente según la ecuación de descarga:
   $$S(t) = S_0 \cdot e^{-t/\tau}$$
   donde $\tau$ representa la constante de depuración homeostática cerebral.

### 9.2. Mecanismos Fisiológicos de Disrupción Tecnológica

El uso tardío de dispositivos provistos de pantallas con retroiluminación LED perturba ambos procesos a través de tres vías identificadas empíricamente (LeBourgeois et al., 2017):

*   **Desplazamiento Temporal del Reposo (*Time-Displacement*):** La interactividad digital posterga deliberadamente el momento de apagar la luz e intentar conciliar el sueño, acortando la ventana física disponible de descanso.
*   **Activación Cognitiva y Neuroemocional (*Pre-Sleep Arousal*):** El procesamiento de información, redes sociales y mensajería activa la corteza prefrontal, elevando la tasa cardíaca y bloqueando la desactivación psicofisiológica requerida para inducir el sueño.
*   **Supresión Melatonérgica por Luz Azul:** Las pantallas emiten radiación en el rango espectral de onda corta ($\lambda \approx 450\text{--}480\text{ nm}$), estimulando los fotopigmentos de melanopsina retinales. Esto suprime la síntesis de melatonina pineal y genera un retraso de fase en el Proceso C.

## 11. Variables en estudio

### 11.1. Definición, Tipo de variable, Codificación de variable, Escala de medición y Codificación de modalidades de respuesta

Para eliminar el sesgo de estimación retrospectiva (donde el informante aproxima erróneamente sus horas dormidas), la investigación adopta una **medición cronológica indirecta**. La duración aproximada del sueño se calcula a partir de la diferencia modular entre la hora de acostarse ($t_{\text{acostarse}}$) y la de levantarse ($t_{\text{levantarse}}$):

$$t_{\text{sueño}} = \begin{cases} 
t_{\text{levantarse}} - t_{\text{acostarse}}, & \text{si } t_{\text{levantarse}} \ge t_{\text{acostarse}} \\[0.2cm]
(t_{\text{levantarse}} + 24) - t_{\text{acostarse}}, & \text{si } t_{\text{levantarse}} < t_{\text{acostarse}}
\end{cases}$$

Este cálculo se desagrega para días laborables lectivos ($t_{\text{sueño, sem}}$) y fines de semana ($t_{\text{sueño, fin}}$), permitiendo cuantificar la compensación o "jetlag social":
$$\Delta t_{\text{jetlag}} = t_{\text{sueño, fin}} - t_{\text{sueño, sem}}$$

```python
# ==============================================================================
# PIPELINE DE PROCESAMIENTO, FRECUENCIAS Y DESCRIPTIVOS (n = 28)
# ==============================================================================
import pandas as pd
import numpy as np
from scipy import stats

ARCHIVO = "../analisis_control_de_datos/Encuesta sobre uso de dispositivos y calidad del sueño (Respuestas).xlsx"
df_raw = pd.read_excel(ARCHIVO).iloc[:28].copy()
df2 = df_raw.drop(columns=["Marca temporal"]).copy()
df2.columns = [f"Pregunta {i}" for i in range(1, 13)]

for c in df2.select_dtypes(include=["object", "string"]).columns:
    df2[c] = df2[c].map(lambda v: v.strip() if isinstance(v, str) else v)

df2["Edad"] = pd.to_numeric(
    df2["Pregunta 2"].astype(str).str.extract(r"(\d+)")[0], errors="coerce"
).astype("int64")

def to_hours(val):
    if hasattr(val, "hour"):
        return val.hour + val.minute / 60.0
    h, m = str(val).split(":")[:2]
    return int(h) + int(m) / 60.0

def fix_bedtime(h):
    h_arr = np.asarray(h, dtype=np.float64)
    return np.where(
        (h_arr >= 8.0) & (h_arr < 12.0),
        h_arr + 12.0,
        np.where((h_arr >= 12.0) & (h_arr < 13.0), h_arr - 12.0, h_arr),
    )

lev_sem = df2["Pregunta 7"].apply(to_hours).to_numpy(dtype=np.float64)
aco_sem = fix_bedtime(df2["Pregunta 8"].apply(to_hours))
lev_fin = df2["Pregunta 9"].apply(to_hours).to_numpy(dtype=np.float64)
aco_fin = fix_bedtime(df2["Pregunta 10"].apply(to_hours))

sem = np.where(aco_sem >= 14.0, (24.0 - aco_sem) + lev_sem, lev_sem - aco_sem)
fin = np.where(aco_fin >= 14.0, (24.0 - aco_fin) + lev_fin, lev_fin - aco_fin)
inconsistente = (df2["Pregunta 8"].apply(to_hours).to_numpy(dtype=np.float64) == lev_sem) | (sem <= 0)
sem[inconsistente] = np.sort(sem[~inconsistente])[0]  # Cota mínima válida = 4.00 h

df2["Sueno_Semana"] = sem
df2["Sueno_Finde"] = fin
df2["Sueno_Ponderado"] = (5.0 * sem + 2.0 * fin) / 7.0
df2["Compensacion_Finde"] = fin - sem

mapa_minutos = {
    "Menos de 30 minutos": 15.0,
    "De 30 a 59 minutos": 44.5,
    "De 60 a 119 minutos": 89.5,
    "120 a 179 minutos": 149.5,
    "180 minutos o más": 195.0,
}
mapa_calidad = {
    "Somnolencia persistente que afecta su concentración en clases.": 1,
    "Cansancio leve que desaparece tras la primera hora.": 2,
    "Energía suficiente para toda la jornada académica.": 3,
}
df2["Minutos_Pantalla"] = df2["Pregunta 5"].map(mapa_minutos)
df2["Calidad_Sueño"] = df2["Pregunta 12"].map(mapa_calidad)

print(f"Observaciones válidas procesadas: n = {len(df2)}")
print("Descriptivos horas de sueño (semana / finde / ponderado / delta):")
print(df2[["Sueno_Semana", "Sueno_Finde", "Sueno_Ponderado", "Compensacion_Finde"]].describe().round(2))
```

### 14.2. Interpretación y Análisis de Resultados

A partir del análisis y procesamiento estadístico de los datos obtenidos de la muestra de 28 estudiantes, se pudieron desprender diversos hallazgos que responden a los objetivos de la investigación, los cuales se presentan a continuación:

1. **Prevalencia de dispositivos y tiempo de exposición:**  
En el Cuadro 1 se observa que el 67,86 % de los estudiantes de la Escuela utilizan teléfonos inteligentes horas previas al descanso. Este porcentaje de estudiantes adquiere diversas repercusiones biológicas por factores como la proximidad del dispositivo a rangos de entre 20 y 30 centímetros de los ojos, lo cual maximiza la irradiancia fototóxica sobre las retinas e inhibe la producción de melatonina en la glándula pineal con mayor severidad comparado con el uso de otros dispositivos como los televisores, los cuales representaron un 14,29 % de uso en horas previas al descanso.  
El Cuadro 2 demuestra que el 57,14 % de los estudiantes supera los 60 minutos de uso continuo de dispositivos electrónicos horas previas al descanso, lo cual abre una ventana a la estimulación lumínica producida por la luz azul, generando sobreestimulación cognitiva en horas previas al reposo.

2. **Déficit lectivo y compensaciones en fines de semana:**  
El Cuadro 3 y el Cuadro 4 presentados anteriormente exponen que el 53,57 % de los estudiantes que respondieron la encuesta se privan del sueño durante los días de mayor actividad académica (de lunes a viernes), en los cuales duermen menos de 7 horas. Se puede observar que, de hecho, los estudiantes promedian 6,67 horas de descanso durante estos días de semana.  
En contraste con lo anteriormente expuesto, durante los días de fines de semana los estudiantes suelen expandir en promedio su descanso compensando ese déficit generado en la semana laboral. En fines de semana estos estudiantes registraron un promedio de 8,93 horas de descanso, lo cual compensa 2,26 horas adicionales a las que dormirían en promedio en días laborales. Este patrón revelado en la muestra expone el denominado comportamiento de *Jetlag Social*, donde el estudiante acumula una deuda de sueño semanal impuesta por los compromisos académicos y laborales (según sea el caso) y la intenta compensar en los días libres.

3. **Disociación cuantitativa y cualitativa:**  
La estimación del modelo de regresión lineal expuesta en el Cuadro 7 y en el Gráfico 12 revela que la pendiente entre los minutos de exposición a las pantallas y la duración del sueño ($\hat{\beta}_1 = -0{,}0063$; $p = 0{,}2326$) no alcanzó la significancia requerida. Sin embargo, al observar la correlación de Spearman en el Cuadro 8 y en el Gráfico 13, se constata que esta arrojó una correlación positiva y estadísticamente significativa ($r_s = +0{,}4064$; $p = 0{,}0318$) respecto al puntaje de calidad percibida al despertar ($1 = \text{somnolencia persistente}$, $2 = \text{cansancio leve}$, $3 = \text{energía suficiente}$). Este hallazgo resulta central para la investigación ya que revela que la exposición a las pantallas de dispositivos electrónicos en los estudiantes no reduce el tiempo de descanso en la cama, o por lo menos no de forma lineal, ya que la hora de despertar de los estudiantes es una variable exógena regida por el horario académico; no obstante, debido al efecto confusor de horarios de despertar muy tempranos ($05\text{:}00\text{--}05\text{:}50\text{ a. m.}$) concentrados en el grupo de menor exposición ($< 60\text{ min}$), el puntaje matutino promedio sube de $1{,}58$ puntos en el tramo $< 60\text{ min}$ a $1{,}90$ en $60\text{--}119\text{ min}$ y $2{,}00$ en $\ge 120\text{ min}$ (revisar Cuadro 5).

#### 14.3.1. Estimación por Intervalos

Con la muestra de 28 estudiantes de la EECA se calcularon intervalos de confianza del 95 % para los parámetros clave (Cuadro 9) empleando la distribución $t$ de Student con $gl = n - 1 = 27$ grados de libertad ($t_{0{,}975;\,27} = 2{,}05183$) debido al tamaño reducido de la muestra.  
La media del sueño en días lectivos arrojó un intervalo de $\mu_{\text{sem}}$ entre $6{,}08$ y $7{,}26$ horas. El límite inferior pone en evidencia el riesgo biológico en la higiene del sueño del estudiantado, aunque el intervalo incluye el umbral clínico de $7{,}00$ horas, en concordancia con el no rechazo de la Prueba 1.  
La media del sueño en fines de semana arrojó un intervalo de $\mu_{\text{fin}}$ entre $8{,}32$ y $9{,}54$ horas, el cual no presenta intersección con el intervalo anterior de los días de semana lectiva; dado que ambas mediciones provienen de los mismos 28 sujetos (estructura pareada), la divergencia entre ambos regímenes se valida formalmente mediante la diferencia individual $\Delta t$.  
La media de compensación horaria ($\mu_{\Delta t}$) se ubicó en un intervalo de $1{,}42$ a $3{,}10$ horas. Al excluir el cero, este intervalo confirma que la muestra estudiantil acumula efectivamente una deuda de sueño durante la semana, la cual es compensada durante los fines de semana.


```python
# ==============================================================================
# ESTIMACIÓN POR INTERVALOS DE CONFIANZA (95 %) Y NORMALIDAD (SHAPIRO-WILK)
# ==============================================================================
def intervalo_media(data, confianza=0.95):
    a = np.asarray(data, dtype=float)
    n = len(a)
    m, se = np.mean(a), stats.sem(a)
    h = se * stats.t.ppf((1.0 + confianza) / 2.0, n - 1)
    return m, m - h, m + h

for etiqueta, col in [
    ("Sueño Semana (h)", "Sueno_Semana"),
    ("Sueño Fin de Semana (h)", "Sueno_Finde"),
    ("Compensación Finde (h)", "Compensacion_Finde"),
]:
    m, li, ls = intervalo_media(df2[col])
    w, p_sw = stats.shapiro(df2[col])
    print(f"{etiqueta}: media = {m:.4f}, IC 95% = [{li:.2f}, {ls:.2f}] | Shapiro-Wilk W = {w:.4f}, p = {p_sw:.4f}")
```

#### 14.3.2. Contraste de Hipótesis

Se realizaron 4 pruebas de contraste de hipótesis en esta investigación ($lpha = 0{,}05$; resumen en el Cuadro 10), las cuales son expuestas a continuación:  

1. **Prueba de Hipótesis 1 correspondiente al déficit de sueño respecto al umbral saludable:**
   $$H_0: \mu_{\text{sem}} \ge 7 \quad \text{vs.} \quad H_1: \mu_{\text{sem}} < 7$$
   Se obtuvo un estadístico $t = -1{,}1559$ con $gl = 27$ y $p = 0{,}1289$ (unilateral izquierdo). Por tal motivo no hay elementos suficientes para rechazar la hipótesis nula. A pesar de que la media muestral es deficitaria ($6{,}67\text{ h}$), la variabilidad en la muestra ($s = 1{,}5124\text{ h}$) y el tamaño reducido de la misma ($n = 28$) limitan la potencia del contraste para certificar el déficit poblacional al 95 % de confianza.

2. **Prueba de Hipótesis 2 sobre asimetría semanal y fenómeno del Jetlag Social:**
   $$H_0: \mu_{\Delta t} \le 0 \quad \text{vs.} \quad H_1: \mu_{\Delta t} > 0$$
   La prueba $t$ pareada arroja un estadístico $t = 5{,}5401$ ($gl = 27$) con $p = 3{,}58 \times 10^{-6}$ unilateral derecho ($p = 7{,}16 \times 10^{-6}$ bilateral), complementada por la prueba no paramétrica de rangos con signo de Wilcoxon ($W = 13{,}0$; $p_{\text{unilateral}} = 1{,}13 \times 10^{-5}$) debido a la no normalidad de $\Delta t$ ($p_{\text{Shapiro-Wilk}} = 0{,}0456$). Esto indica que se rechaza la hipótesis nula; por lo tanto, existe una compensación horaria estadísticamente significativa en el descanso durante los fines de semana.

3. **Prueba de Hipótesis 3 sobre el modelo lineal (MCO) e independencia en tablas de contingencia:**
   *Modelo lineal (Cuadro 7):*
   $$H_0: \beta_1 = 0 \quad \text{vs.} \quad H_1: \beta_1 \neq 0$$
   Con un estadístico $F(1, 26) = 1{,}494$ ($t = -1{,}2221$; $p = 0{,}2326$; $R^2 = 0{,}0543$; $R^2_{\text{aj}} = 0{,}0180$) no hubo elementos suficientes para rechazar la hipótesis nula, por lo que no se puede confirmar que exista una relación lineal directa entre los minutos de exposición a pantallas de dispositivos electrónicos y la duración del descanso semanal.  
   *Independencia en contingencia (Cuadro 5):*
   $$H_0: \text{El tramo de tiempo de pantalla y la sensación al despertar son independientes}$$
   El contraste chi-cuadrado sobre la tabla $3 \times 3$ arroja $\chi^2 = 5{,}7100$ ($gl = 4$; $p = 0{,}2219$), y $\chi^2 = 7{,}2657$ ($gl = 8$; $p = 0{,}5083$) en la matriz $5 \times 3$, no rechazándose tampoco $H_0$ (con $77{,}8\,\%$ de frecuencias esperadas menores que $5$ en el cruce $3 \times 3$).

4. **Prueba de Hipótesis 4 sobre correlación lineal y monótona:**
   $$H_0: \rho = 0 \quad \text{vs.} \quad H_1: \rho \neq 0 \quad (gl = n - 2 = 26)$$
   La correlación lineal de Pearson entre minutos de pantalla y horas de sueño semanal resultó no significativa ($r = -0{,}2331$; $p = 0{,}2326$), al igual que la correlación de Spearman frente a los despertares nocturnos ($r_s = -0{,}0011$; $p = 0{,}9957$). En cambio, para la calidad percibida al despertar ($1 = \text{somnolencia}$, $2 = \text{cansancio leve}$, $3 = \text{energía}$), Spearman arroja $r_s = +0{,}4064$ y $p = 0{,}0318$, por lo que se rechaza la hipótesis nula; esto confirma una asociación monótona estadísticamente significativa en esta muestra, aunque de signo positivo debido al sesgo confusor de los estudiantes que despiertan entre las $05\text{:}00$ y las $05\text{:}50\text{ a. m.}$ en el grupo de menor exposición (Cuadro 8).


```python
# ==============================================================================
# MÓDULO DE PRUEBAS DE HIPÓTESIS Y CORRELACIONES (n = 28)
# ==============================================================================
import statsmodels.api as sm

t1 = stats.ttest_1samp(df2["Sueno_Semana"], popmean=7.0, alternative="less")
t2 = stats.ttest_rel(df2["Sueno_Finde"], df2["Sueno_Semana"], alternative="greater")
w2 = stats.wilcoxon(df2["Sueno_Finde"], df2["Sueno_Semana"], alternative="greater")

X = sm.add_constant(df2["Minutos_Pantalla"])
modelo = sm.OLS(df2["Sueno_Semana"], X).fit()

ct_3x3 = pd.crosstab(
    df2["Pregunta 5"].map({
        "Menos de 30 minutos": "< 60 min",
        "De 30 a 59 minutos": "< 60 min",
        "De 60 a 119 minutos": "60 a 119 min",
        "120 a 179 minutos": "120 min o más",
        "180 minutos o más": "120 min o más",
    }),
    df2["Calidad_Sueño"],
)
chi2_val, p_chi2, dof_chi2, _ = stats.chi2_contingency(ct_3x3)

r_pear, p_pear = stats.pearsonr(df2["Minutos_Pantalla"], df2["Sueno_Semana"])
r_sp_cal, p_sp_cal = stats.spearmanr(df2["Minutos_Pantalla"], df2["Calidad_Sueño"])
desp_num = df2["Pregunta 11"].map({"Ninguna": 0, "Una vez": 1, "Dos veces": 2, "3 o más veces": 3})
r_sp_desp, p_sp_desp = stats.spearmanr(df2["Minutos_Pantalla"], desp_num)

print(f"Prueba 1 (mu_sem < 7): t = {t1.statistic:.4f}, gl = 27, p = {t1.pvalue:.4f}")
print(f"Prueba 2 (delta > 0): t = {t2.statistic:.4f}, gl = 27, p_unilateral = {t2.pvalue:.4e} | Wilcoxon W = {w2.statistic:.1f}, p_uni = {w2.pvalue:.4e}")
print(f"Prueba 3a (OLS): F(1, 26) = {modelo.fvalue:.4f}, p = {modelo.f_pvalue:.4f}, R2 = {modelo.rsquared:.4f}, R2_aj = {modelo.rsquared_adj:.4f}")
print(f"Prueba 3b (Chi2 3x3): chi2 = {chi2_val:.4f}, gl = {dof_chi2}, p = {p_chi2:.4f}")
print(f"Prueba 4 (Pearson / Spearman, gl = 26): r = {r_pear:.4f} (p={p_pear:.4f}), rs_cal = {r_sp_cal:+.4f} (p={p_sp_cal:.4f}), rs_desp = {r_sp_desp:.4f} (p={p_sp_desp:.4f})")
```

## 15. Dificultades Encontradas en la Investigación

La investigación planteada enfrentó diversas limitaciones tanto metodológicas como operativas y contextuales, las cuales se detallan a continuación:

1. **Sesgo de la memoria e inexactitud del autorreporte:**  
Al depender de un cuestionario digital de autorreporte, las estimaciones horarias suministradas por los estudiantes de la EECA están expuestas a la imprecisión de la memoria. Es común que los encuestados confundan el tiempo de permanencia en la cama con las horas reales de descanso, por lo que obvian la brecha existente entre ambos períodos.

2. **Tamaño de la muestra y participación:**  
La participación en el autorreporte fue de tan solo 28 encuestados válidos, número que limita la potencia estadística de los contrastes de hipótesis y amplía los intervalos de confianza, incrementando así el margen de error estándar en las pruebas paramétricas respecto a efectos moderados o relaciones no lineales.

3. **Muestreo no probabilístico:**  
La difusión del instrumento vía mensajería introduce un sesgo de selección, pues los participantes debían pertenecer a determinados grupos estudiantiles para acceder al cuestionario, por lo que no se alcanzó la totalidad poblacional esperada.

4. **Diversas variables exógenas:**  
Factores como el consumo de cafeína o bebidas energizantes en distintos momentos del día, los tiempos de traslado en transporte público, las cargas laborales y otras actividades extracurriculares afectan la rutina estudiantil e introducen ruido estadístico que incide directamente sobre las horas de reposo y la fatiga matutina.

5. **Ambigüedad en la captura horaria y registro inconsistente (Diagnóstico de Calidad de Datos):**  
El selector de horas de Google Forms generó confusión entre formatos de 12 y 24 horas en los registros de acostarse cercanos a las 12:00, requiriendo desplazamiento programático ($+12\text{ h}$ entre `08:00` y `11:59`; $-12\text{ h}$ entre `12:00` y `12:59`) y exclusión de las horas brutas de reloj en las gráficas descriptivas. Adicionalmente, un registro declaró idéntica hora de acostarse y levantarse (`09:00`–`09:00`) junto a un uso nocturno $\ge 180\text{ min}$, obligando a su imputación mediante la cota mínima válida de la muestra ($4{,}00\text{ h}$), mientras que `Edad` requirió depuración de texto libre (`"20 años"`).

6. **Ajuste operativo respecto al anteproyecto:**  
El formulario definitivo aplicado omitió las variables *Carrera cursada* y *Latencia pre-sueño* proyectadas originalmente, adoptó cuatro niveles en la escala de sensación matutina (en lugar de cinco) y cerró el corte analítico en $n = 28$ observaciones anteriores al procesamiento diagnóstico (excluyendo una respuesta tardía del 15 de noviembre de 2026).

Acorde con los objetivos de investigación planteados y los métodos de estimación utilizados, se llega a las siguientes conclusiones:

### Anexo C: Glosario de Términos

**Correlación de Pearson ($r$):** Medida paramétrica de asociación lineal entre dos variables cuantitativas continuas. Su magnitud oscila en $[-1, 1]$ y evalúa únicamente proporcionalidad rectilínea, perdiendo sensibilidad frente a relaciones monótonas curvas o escalas ordinales.

**Correlación de Spearman ($r_s$):** Coeficiente no paramétrico calculado sobre los rangos ordenados de dos variables ($gl = n - 2$). Resulta insensible a la asimetría distribucional y a valores extremos, operando válidamente sobre categorías ordinales como la percepción matutina al despertar.

**Higiene del sueño:** Conjunto de hábitos conductuales y condiciones ambientales orientados a estabilizar la fase circadiana y preservar la continuidad del descanso nocturno. Incluye la desconexión lumínica previa al reposo, la regularidad entre días lectivos y fines de semana y la reducción de estímulos cognitivos en la cama; su incumplimiento se refleja en latencias prolongadas, fragmentación nocturna y acumulación de deuda semanal.

**Imputación por cota mínima:** Regla de saneamiento que sustituye un registro cronológicamente inviable por el menor valor empíricamente válido observado dentro de la muestra depurada. Frente a la eliminación del caso, conserva el tamaño muestral $n = 28$ sin introducir valores extremos artificiales ($0{,}0\text{ h}$ o $12{,}0\text{ h}$) que distorsionarían la varianza muestral.

**Jetlag social ($\Delta t$):** Discrepancia horaria entre la duración o fase del sueño en días libres frente a los días lectivos o laborales ($\Delta t = t_{\text{fin}} - t_{\text{sem}}$). Cuantifica el desajuste impuesto por el horario académico temprano sobre el marcapasos circadiano endógeno del estudiante.

**Latencia del sueño:** Intervalo temporal medido desde el instante en que el individuo apaga la fuente lumínica e intenta dormir hasta el inicio efectivo de la primera fase no REM. La activación cortical nocturna y la supresión hormonal alargan esta transición, reduciendo el tiempo útil de reposo aunque la permanencia total en cama parezca constante.

**Luz azul:** Fracción del espectro visible de onda corta comprendida aproximadamente entre $450$ y $480\text{ nm}$, emitida de forma predominante por diodos LED de teléfonos inteligentes y monitores. Al incidir sobre la retina a corta distancia, excita las células ganglionares con melanopsina y frena la síntesis pineal de melatonina, desplazando la señal de inicio del sueño (Proceso C).

**Mínimos Cuadrados Ordinarios (OLS / MCO):** Estimador lineal que fija el intercepto $\hat{\beta}_0$ y la pendiente $\hat{\beta}_1$ minimizando la suma de los cuadrados de las discrepancias verticales entre los valores observados y la recta ajustada ($\sum e_i^2$). Bajo homocedasticidad, independencia de errores y normalidad residual, produce los estimadores lineales insesgados de varianza mínima; en muestras pequeñas ($n = 28$), su potencia estadística $1 - \beta$ solo detecta efectos de magnitud pronunciada.

**Potencia estadística ($1 - \beta$):** Probabilidad de rechazar una hipótesis nula falsa dado un tamaño muestral $n$, una varianza residual $\sigma^2$ y un nivel de significancia $\alpha$. Con $n = 28$, efectos débiles sobre la duración del descanso quedan por debajo del umbral de detección del contraste $t$ o $F$.

**Proceso C y Proceso S:** Componentes acoplados del modelo regulador de Borbély (1982). El Proceso C describe la oscilación circadiana endógena de período cercano a $24\text{ h}$ sincronizada por la luz ambiental, mientras que el Proceso S mide la presión homeostática acumulativa de sueño generada durante la vigilia continua.

**Prueba de Shapiro-Wilk ($W$):** Contraste formal de bondad de ajuste que evalúa la hipótesis nula de que una muestra proviene de una población normalmente distribuida. El estadístico $W \in (0, 1]$ compara los cuantiles ordenados muestrales contra los cuantiles teóricos normales; valores bajos de $W$ con $p < 0{,}05$ —como en $\Delta t$ ($W = 0{,}9248; p = 0{,}0456$)— rechazan la normalidad y exigen complementar las pruebas $t$ con alternativas no paramétricas como Wilcoxon.

**Prueba de rangos con signo de Wilcoxon:** Contraste pareado no paramétrico basado en la suma de rangos de las diferencias absolutas según su signo. Prescinde del supuesto de normalidad sobre $\Delta t$, verificando si la mediana de compensación semanal difiere sistemáticamente de cero.

**Sesgo de confusión:** Distorsión de la asociación observada entre exposición y desenlace provocada por una tercera variable vinculada a ambos fenómenos. En esta muestra, los horarios de despertar muy tempranos ($05\text{:}00\text{--}05\text{:}50\text{ a. m.}$) se concentran en el tramo de menor uso de pantallas, generando una correlación positiva espuria entre minutos de pantalla y bienestar matutino.
