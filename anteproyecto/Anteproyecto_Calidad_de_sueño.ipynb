## 2. Planteamiento del Problema

Desde la popularización y adopción masiva de la tecnología móvil y los terminales digitales portátiles, se ha producido una transformación sustancial en las rutinas de la población a nivel global (LeBourgeois et al., 2017). Esta dinámica adquiere especial relevancia en la población estudiantil universitaria, cuyos integrantes integran de forma intensiva el uso de teléfonos inteligentes, computadoras portátiles y tabletas en sus actividades diurnas y nocturnas.

La interacción recurrente con estos terminales en horarios nocturnos altera la calidad del descanso mediante mecanismos fisiológicos y conductuales ampliamente documentados (LeBourgeois et al., 2017): la sustitución del tiempo destinado a dormir por el consumo de contenidos digitales, la estimulación neurocognitiva derivada de las pantallas interactivas y la exposición directa a la luz azul, la cual inhibe la secreción de melatonina y altera los ritmos circadianos. Esta situación resulta alarmante considerando que el descanso regular y adecuado constituye un pilar indispensable para el rendimiento intelectual y la salud integral (Hirshkowitz et al., 2015). A escala mundial, cerca de un tercio de la población adulta duerme por debajo del rango óptimo de 7 a 9 horas diarias (Liu et al., 2016).

En el ámbito particular de la Escuela de Estadística y Ciencias Actuariales de la Universidad Central de Venezuela (EECA-UCV), los planes de estudio exigen una alta carga cognitiva, un elevado esfuerzo de abstracción cuantitativa y jornadas prolongadas de dedicación frente al computador en tareas de programación, cálculo actuarial y análisis de datos. En el transcurso del período académico 2026-01S, la comunidad estudiantil asume compromisos académicos y dinámicas de transporte urbano que desplazan frecuentemente las horas de lectura, estudio e interacción tecnológica hacia la noche tardía.

A pesar de que el cansancio matutino se manifiesta de manera cotidiana en las aulas de clase, la EECA-UCV no dispone de un relevamiento métrico formal que describa la persistencia de estos hábitos y cuantifique la asociación empírica entre la exposición tecnológica previa al reposo y el déficit de sueño en su alumnado. En consideración a este vacío de información, se formula la siguiente pregunta central de investigación:

> ### **Pregunta de Investigación:**
> **¿De qué manera la duración de la exposición nocturna a pantallas de dispositivos electrónicos altera la calidad y las horas efectivas de sueño en los estudiantes de la EECA-UCV?**

## 3. Objetivos

### 3.1. Objetivo(s) General(es)

Analizar la relación lineal entre el tiempo de uso nocturno de dispositivos electrónicos y la calidad del sueño en los estudiantes de la EECA-UCV durante el período 2026-01S.

### 3.2. Objetivos Específicos

1. **Caracterizar** los hábitos de consumo digital nocturno según el tipo de dispositivo y la ventana de exposición previa al descanso.
2. **Estimar** la duración absoluta y la calidad percibida del sueño durante los días lectivos frente a los fines de semana.
3. **Identificar** las variaciones en los patrones de descanso asociadas a los distintos niveles de exposición a pantallas.


---

## 4. Finalidad

La recolección de estos datos primarios permite dimensionar un problema latente en la dinámica universitaria. Cuantificar el déficit de sueño frente a la exposición digital proporciona la base métrica empírica necesaria para que las autoridades de la Escuela estructuren campañas de higiene del sueño fundamentadas en su propia población, sustituyendo las suposiciones genéricas sobre el cansancio estudiantil.

## 5. Justificación

El descanso nocturno no constituye un simple intervalo de inactividad, sino un proceso neurobiológico activo indispensable para la consolidación sináptica, la memoria de trabajo y la recuperación metabólica (Hirshkowitz et al., 2015). En el ámbito de la EECA-UCV, donde el rendimiento del estudiante se basa principalmente en mantener una concentración sostenida por largos períodos de tiempo frente a asignaturas de alta exigencia analítica (como en Matemáticas, Estadísticas, Probabilidades, etc.), la privación de sueño deteriora directamente la agilidad mental de los estudiantes y eleva el margen de error en el razonamiento cuantitativo, principalmente durante semanas de alta carga académica.

A nivel metodológico, la investigación deja a la comunidad académica un instrumento de medición validado para la realidad del estudiante universitario venezolano. El diseño estructural del cuestionario y la operacionalización de las variables establecen un precedente referencial. Futuros estudios observacionales en otras facultades podrán replicar esta arquitectura para contrastar el fenómeno del "jetlag social" y la disrupción circadiana en diferentes disciplinas.

## 8. Antecedentes

El marco empírico de esta investigación se fundamenta en estudios internacionales recientes de alta rigurosidad científica que han examinado la relación entre la tecnología digital y los parámetros de descanso:

* **He, X., Pan, B., Ma, N., Li, D., Kong, W., Liu, Q., Liu, X., Wang, X., Deng, X., & Yang, K. (2025):**  
  *The association of screen time and the risk of sleep outcomes: A systematic review and meta-analysis*. Publicado en *Frontiers in Psychiatry*, *16*, 1640263.  
  Este trabajo representa un aporte metodológico directo al sintetizar 21 estudios de cohorte con 548.000 participantes. Los autores cuantificaron el riesgo relativo y la reducción marginal de minutos de sueño mediante modelos de metarregresión, lo cual fundamenta nuestra operacionalización del tiempo de exposición en intervalos continuos y la evaluación del riesgo de insomnio matutino.

* **Hysing, M., Pallesen, S., Stormark, K. M., Jakobsen, R., Lundervold, A. J., & Sivertsen, B. (2015):**  
  *Sleep and use of electronic devices in adolescence: Results from a large population-based study*. Publicado en *BMJ Open*, *5*(1), e006748.  
  Esta investigación constituye un precedente metodológico clave sobre el autorreporte de dispositivos en jóvenes. Los autores emplearon modelos de regresión logística ajustados por variables sociodemográficas para estimar el déficit de sueño, lo cual justifica nuestra estrategia de medir la latencia pre-sueño y discriminar el uso según el tipo de pantalla interactiva.

* **LeBourgeois, M. K., Hale, L., Chang, A. M., Akacem, L. D., Montgomery-Downs, H. E., & Buxton, O. M. (2017):**  
  *Digital media and sleep in childhood and adolescence*. Publicado en *Pediatrics*, *140*(Suppl 2), S92--S96.  
  Este estudio aporta la estructura conceptual para aislar los mecanismos fisiológicos del descanso. Los autores categorizan el desplazamiento temporal, la activación psicológica y la luz azul en el modelo de dos procesos, sustentando nuestra medición diferenciada de los horarios de acostarse frente a las horas efectivas de sueño.

* **Maurya, C., Muhammad, T., Maurya, P., & Dhillon, P. (2022):**  
  *The association of smartphone screen time with sleep problems among adolescents and young adults: Cross-sectional findings from India*. Publicado en *BMC Public Health*, *22*(1), 1686.  
  El aporte metodológico de este trabajo radica en su diseño transversal con jóvenes universitarios. Los autores aplicaron pruebas de chi-cuadrado y razones de prevalencia estratificadas para evaluar el umbral crítico de tres horas, lo cual orienta la construcción de nuestras tablas de contingencia y el análisis bivariado de datos.

* **Zhong, C., Masters, M., Donzella, S. M., Diver, W. R., & Patel, A. V. (2025):**  
  *Electronic screen use and sleep duration and timing in adults*. Publicado en *JAMA Network Open*, *8*(3), e252493.  
  Este antecedente fundamenta el análisis de la ventana temporal previa a la conciliación. Los autores modelaron el uso de pantallas en la última hora antes de dormir mediante ecuaciones de estimación generalizada, lo que respalda nuestra decisión de restringir la ventana crítica de exposición a las dos horas pre-descanso.

## 9. Marco teórico referencial o bases teóricas

### 9.1. Modelo de Dos Procesos de Regulación del Sueño (Borbély, 1982)

La fisiología del descanso humano está gobernada por la interacción dinámica de dos fuerzas reguladoras:

1. **Proceso C (Regulación Circadiana):**  
   Constituye una oscilación endógena biológica cuasi-sinusoidal con un período aproximado de 24 horas, generada por el marcapasos central ubicado en el núcleo supraquiasmático (NSQ) del hipotálamo anterior. Este mecanismo sincroniza las ventanas de alerta y propensión al sueño con las señales lumínicas ambientales captadas por las células ganglionares de la retina.

2. **Proceso S (Presión Homeostática):**  
   Representa la acumulación neuroquímica progresiva de propensión al descanso (mediada por la adenosina en el prosencéfalo basal) generada por cada hora de vigilia continua. Durante el sueño no REM, el Proceso S decae exponencialmente según la ecuación de descarga:
   $$S(t) = S_0 \cdot e^{-t/\tau}$$
   donde $\tau$ representa la constante de depuración homeostática cerebral.

### 9.2. Mecanismos Fisiológicos de Disrupción Tecnológica

El uso tardío de dispositivos provistos de pantallas con retroiluminación LED perturba ambos procesos a través de tres vías identificadas empíricamente (LeBourgeois et al., 2017):

*   **Desplazamiento Temporal del Reposo (*Time-Displacement*):** La interactividad digital posterga deliberadamente el momento de apagar la luz e intentar conciliar el sueño, acortando la ventana física disponible de descanso.
*   **Activación Cognitiva y Neuroemocional (*Pre-Sleep Arousal*):** El procesamiento de información, redes sociales y mensajería activa la corteza prefrontal, elevando la tasa cardíaca y bloqueando la desactivación psicofisiológica requerida para inducir el sueño.
*   **Supresión Melatonérgica por Luz Azul:** Las pantallas emiten radiación en el rango espectral de onda corta ($\lambda \approx 450\text{--}480\text{ nm}$), estimulando los fotopigmentos de melanopsina retinales. Esto suprime la síntesis de melatonina pineal y genera un retraso de fase en el Proceso C.

## 11. Variables en estudio

### 11.1. Definición, Tipo de variable, Codificación de variable, Escala de medición y Codificación de modalidades de respuesta

Para eliminar el sesgo de estimación retrospectiva (donde el informante aproxima erróneamente sus horas dormidas), la investigación adopta una **medición cronológica indirecta**. La duración aproximada del sueño se calcula a partir de la diferencia modular entre la hora de acostarse ($t_{\text{acostarse}}$) y la de levantarse ($t_{\text{levantarse}}$):

$$t_{\text{sueño}} = \begin{cases} 
t_{\text{levantarse}} - t_{\text{acostarse}}, & \text{si } t_{\text{levantarse}} \ge t_{\text{acostarse}} \\[0.2cm]
(t_{\text{levantarse}} + 24) - t_{\text{acostarse}}, & \text{si } t_{\text{levantarse}} < t_{\text{acostarse}}
\end{cases}$$

Este cálculo se desagrega para días laborables lectivos ($t_{\text{sueño, sem}}$) y fines de semana ($t_{\text{sueño, fin}}$), permitiendo cuantificar la compensación o "jetlag social":
$$\Delta t_{\text{jetlag}} = t_{\text{sueño, fin}} - t_{\text{sueño, sem}}$$

## 15. Posibles Dificultades para la Realización del Trabajo

1. **Inexactitud en el autorreporte:** Al ser un cuestionario retrospectivo, los minutos y horas reportadas dependen de la memoria del encuestado.
2. **Confusión en formatos horarios:** Riesgo de errores al ingresar horas en formato 12h vs 24h, subsanado mediante validaciones en el formulario.
3. **Tasa de respuesta voluntaria:** Posible renuencia o apatía estudiantil durante períodos de exámenes, mitigada mediante difusión activa por delegados.
4. **Variables confusoras:** Hábitos como consumo de café o jornadas laborales nocturnas que requieren análisis cualitativo adicional.
